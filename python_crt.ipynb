```python
a=input("Enter the password:")
up=0
lc=0
sp=0
d=0
if len(a)>=8:
    for i in a:
        if i.isupper():
            up=up+1
        elif i.islower():
            lc=lc+1
        elif i.isdigit():
            d=d+1
        else:
            sp=sp+1
    if up>0 and lc>0 and sp>0 and d>0:
        print("password verified")
    else:
        print("please enter correct password")
else:
    print("please enter correct password")
```

```python
#Round Robin Tournament Scheduling
import random
a=int(input("Enter number of Matches:"))
teams=[]
print("Enter all teams Names:")
for i in range(a):
    name=input()
    teams.append(name)
matches=[]

for i in range(0,a-1):
    for j in range(i+1,a):
        matches.append([teams[i],teams[j]])
random.shuffle(matches)
pos=1
for i in matches:
    print("Match {} : {} vs {}".format(pos,i[0],i[1]))
    pos=pos+1
```

```python
a=input("Enter Boy's Name:")
b=input("Enter Girls's Name:")
a1=list(a)
b1=list(b)
for i in range(len(a1)):
    for j in range(len(b1)):
        if a1[i]==b1[j]:
            a1[i]==0
            b1[j]==0
cnt=0
for i in a1:
    if i!='0':
        cnt=cnt+1
for i in b1:
    if i!='0':
        cnt=cnt+1
index=0
res=list("FLAMES")
for i in range(5):
    index=(index+(cnt+1))%len(res)
    res.pop(index)
print(res[0])
```

```python
#Pattern program
a=[[" " for i in range(11)] for i in range(11)]
for i in range(11):
    for j in range(11):
        if j==0 or j==10:
            a[i][j]="*"
        if i==j and i<6:
            a[i][j]="*"
        if i+j==10 and i<6:
            a[i][j]="*"

for i in range(11):
    for j in range(11):
        print(a[i][j], end="")
    print()
```
