```python
#Round Robin Tournament Scheduling
import random
a=int(input("Enter number of Matches:"))
teams=[]
print("Enter all teams Names:")
for i in range(a):
    name=input()
    teams.append(name)
matches=[]

for i in range(0,a-1):
    for j in range(i+1,a):
        matches.append([teams[i],teams[j]])
random.shuffle(matches)
pos=1
for i in matches:
    print("Match {} : {} vs {}".format(pos,i[0],i[1]))
    pos=pos+1
```

```python
#Menu-Driven Area Calculator
def Square():
    side = int(input("Enter side: "))
    print("Area of Square =", side * side)

def Rectangle(l, b):
    print("Area of Rectangle =", l * b)

def Circle():
    r = int(input("Enter radius: "))
    print("Area of Circle =", 3.14 * r * r)

def Triangle():
    b = int(input("Enter base: "))
    h = int(input("Enter height: "))
    print("Area of Triangle =", 0.5 * b * h)
def menu():
    while True:
        print("1. Square\n2. Rectangle\n3. Circle\n4. Triangle\n5. Exit\n")
        a = input("Enter your choice: ")

        if a == "1":
            Square()

        elif a == "2":
            l = int(input("Enter length: "))
            b = int(input("Enter breadth: "))
            Rectangle(l, b)
        elif a == "3":
            Circle()
        elif a == "4":
            Triangle()
        elif a == "5":
            print("Exiting...")
            break
        else:
            print("Invalid choice!")

menu()
```

```python
#Special Prime Number Finder
def prime(n):
    if n<=1:
        return False
    else:
        for i in range(2,n):
            if n % i == 0:
                return False
        return True
def sod(n):
    d= [int(i) for i in str(n)]
    return sum(d)
def every(n):
    for i in range(3):
        d = n%10
        n = n//10
        if not prime(d):
            return False
    return True
for i in range(100,1000):
    if prime(i) and prime(sod(i)) and every(i):
        print(i, end=" ")
```

```python
#Run-Length Encoding
a="AAABBCCDDDD"
res=""
c=1
for i in range(len(a)):
    if i+1 <len(a) and a[i]==a[i+1]:
        c=c+1
    else:
        res=res+str(c)
        res=res+a[i]
        c=1
print(res)
```
