```python
a=input("Enter the password:")
up=0
lc=0
sp=0
d=0
if len(a)>=8:
    for i in a:
        if i.isupper():
            up=up+1
        elif i.islower():
            lc=lc+1
        elif i.isdigit():
            d=d+1
        else:
            sp=sp+1
    if up>0 and lc>0 and sp>0 and d>0:
        print("password verified")
    else:
        print("please enter correct password")
else:
    print("please enter correct password")
```

```python
#Round Robin Tournament Scheduling
import random
a=int(input("Enter no of matches:"))
teams=[]
print("Enter all teams names:")
for i in range(a):
    name=input()
    teams.append(name)
matches=[]

for i in range(0,a-1):
    for j in range(i+1,a):
        matches.append([teams[i],teams[j]])
random.shuffle(matches)
pos=1
for i in matches:
    print("Match {} : {} vs {}".format(pos,i[0],i[1]))
    pos=pos+1
```
