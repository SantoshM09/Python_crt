```python
#Round Robin Tournament Scheduling
import random
a=int(input("Enter number of Matches:"))
teams=[]
print("Enter all teams Names:")
for i in range(a):
    name=input()
    teams.append(name)
matches=[]

for i in range(0,a-1):
    for j in range(i+1,a):
        matches.append([teams[i],teams[j]])
random.shuffle(matches)
pos=1
for i in matches:
    print("Match {} : {} vs {}".format(pos,i[0],i[1]))
    pos=pos+1
```

```python
#Menu-Driven Area Calculator
def Square():
    side = int(input("Enter side: "))
    print("Area of Square =", side * side)

def Rectangle(l, b):
    print("Area of Rectangle =", l * b)

def Circle():
    r = int(input("Enter radius: "))
    print("Area of Circle =", 3.14 * r * r)

def Triangle():
    b = int(input("Enter base: "))
    h = int(input("Enter height: "))
    print("Area of Triangle =", 0.5 * b * h)
def menu():
    while True:
        print("1. Square\n2. Rectangle\n3. Circle\n4. Triangle\n5. Exit\n")
        a = input("Enter your choice: ")

        if a == "1":
            Square()

        elif a == "2":
            l = int(input("Enter length: "))
            b = int(input("Enter breadth: "))
            Rectangle(l, b)
        elif a == "3":
            Circle()
        elif a == "4":
            Triangle()
        elif a == "5":
            print("Exiting...")
            break
        else:
            print("Invalid choice!")

menu()
```

```python
#Special Prime Number Finder
def prime(n):
    if n<=1:
        return False
    else:
        for i in range(2,n):
            if n % i == 0:
                return False
        return True
def sod(n):
    d= [int(i) for i in str(n)]
    return sum(d)
def every(n):
    for i in range(3):
        d = n%10
        n = n//10
        if not prime(d):
            return False
    return True
for i in range(100,1000):
    if prime(i) and prime(sod(i)) and every(i):
        print(i, end=" ")
```

```python
#Linked List
class Node:
    def __init__ (self,data):
        self.data=data
        self.next=None
class LinkedList:
    def __init__(self):
        self.head=None
    def insert(self,data):
        newnode=Node(data)
        if self.head==None:
            self.head=newnode
        else:
            temp=self.head
            while temp.next:
                temp=temp.next
            temp.next=newnode

    def display(self):
        temp=self.head
        while temp.next:
            print(temp.data,end="->")
            temp=temp.next


l=LinkedList()
l.insert(100)
l.insert(200)
l.insert(300)
l.insert(400)
l.insert(500)
l.display()
```

### Creating a Web-Based UI with Gradio

Since Tkinter doesn't work in Colab, we can use libraries like `gradio` to create interactive web-based UIs. Gradio allows you to quickly build and share web interfaces for your Python code.

```python
!pip install gradio
```

```python
!pip install googletrans
import gradio as gr
from googletrans import Translator

translator = Translator()

def translate_function(text_to_translate, target_language):
    # Gradio functions are synchronous, so we don't need asyncio.run() or await here
    # However, googletrans's translate method might still be async under the hood
    # For simplicity, we'll keep the direct call. If issues arise, an async wrapper would be needed.
    try:
        translated_text = translator.translate(text_to_translate, dest=target_language).text
        return translated_text
    except Exception as e:
        return f"Error: {e}"


# Create the Gradio interface
iface = gr.Interface(
    fn=translate_function,
    inputs=[
        gr.Textbox(label="Text to Translate"),
        gr.Textbox(label="Target Language (e.g., 'es' for Spanish, 'fr' for French)")
    ],
    outputs=gr.Textbox(label="Translated Text"),
    title="Simple Translation App (Gradio)",
    description="Enter text and a target language to get a translation."
)

# Launch the interface
# Set share=True to get a public link if you want to share it outside Colab
iface.launch()
```

This code will launch a simple web interface within your Colab output, allowing you to input text and a target language, and see the translation. You'll get a local URL and potentially a public URL (if `share=True` is used) to access the UI.

### Creating a Gradio Calculator Interface

```python
import gradio as gr

def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    if b == 0:
        return "Cannot divide by zero!"
    return a / b


# Create the Gradio interface for addition
add_interface = gr.Interface(
    fn=add,
    inputs=["number", "number"],
    outputs="number",
    title="Addition Calculator",
    description="Enter two numbers to add them."
)

# Create the Gradio interface for subtraction
subtract_interface = gr.Interface(
    fn=subtract,
    inputs=["number", "number"],
    outputs="number",
    title="Subtraction Calculator",
    description="Enter two numbers to subtract the second from the first."
)

# Create the Gradio interface for multiplication
multiply_interface = gr.Interface(
    fn=multiply,
    inputs=["number", "number"],
    outputs="number",
    title="Multiplication Calculator",
    description="Enter two numbers to multiply them."
)

# Create the Gradio interface for division
divide_interface = gr.Interface(
    fn=divide,
    inputs=["number", "number"],
    outputs="number",
    title="Division Calculator",
    description="Enter two numbers to divide the first by the second."
)

# Combine interfaces into a TabbedInterface
demo = gr.TabbedInterface(
    [add_interface, subtract_interface, multiply_interface, divide_interface],
    ["Add", "Subtract", "Multiply", "Divide"]
)

# Launch the combined interface
demo.launch()
```

```python
class Graph:
    def __init__ (self,v):
        self.v=v
        self.adj=[[] for _ in range(v)]

    def add_edge(self,a,b):
        self.adj[a].append(b)
        self.adj[b].append(a)

    def display(self):
        for i in range(len(self.adj)):
            print(f"Node{i}:", end=" ")
            for j in self.adj[i]:
                print(j,end="-->")
            print()


a=Graph(6)
a.add_edge(0,1)
a.add_edge(0,3)
a.add_edge(1,4)
a.add_edge(3,2)
a.add_edge(3,3)
a.add_edge(5,1)
a.display()
```
